```python
unique_vals = data1['label'].unique()
len(unique_vals)
```

```python
unique_vals2 = data2['label'].unique()
len(unique_vals2)
```

```python
list1=[]
list2=[]
for i in range(len(unique_vals)):
    for j in range(len(unique_vals2)):
        # print(unique_vals2[j])
        if(unique_vals2[j] == unique_vals[i]):
            list1.append(unique_vals[i]) 
            list2.append(unique_vals2[j])
```

```python
print(list1)
print(list2)
```

```python
data1['label'] = (
    data1['label']
    .str.lower()
    .str.strip()
    .str.replace(r'\s+', ' ', regex=True)
)

data2['label'] = (
    data2['label']
    .str.lower()
    .str.strip()
    .str.replace(r'\s+', ' ', regex=True)
)
```

```python
df = pd.concat([data1, data2], axis=0, ignore_index=True)
```
